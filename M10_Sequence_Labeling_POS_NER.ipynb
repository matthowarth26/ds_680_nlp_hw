## Part of Speech Tagging (POS Tagging)

Text for POS Tagging

```python
pos_text = """
Luke Skywalker is a young farm boy who discovers that he has a powerful connection to the Force.
After meeting Obi-Wan Kenobi, Luke begins training to become a Jedi and joins the Rebel Alliance in its fight against the Galactic Empire.
Meanwhile, the ruthless Darth Vader searches relentlessly for the rebels as they secretly plan to destroy the Empire's dangerous Death Star.
With help from Princess Leia and Han Solo, Luke bravely confronts the enemy and successfully destroys the massive space station.
Although the Empire remains a serious threat, Luke and his friends continue their courageous struggle to restore peace and freedom to the galaxy.
"""
```

Import Libraries

```python
import nltk

nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
```

Tokenize the text

```python
tokens = nltk.word_tokenize(pos_text)
```

Apply POS tagging

```python
pos_tags = nltk.pos_tag(tokens)
```

Display tokens and tags

```python
print(f"{'Word':<20} {'POS Tag':<10}")
print("-" * 30)

for word, tag in pos_tags:
    print(f"{word:<20} {tag:<10}")
```

## Named Entity Recognition (NER)

Text for NER

```python
ner_text = """
Anthropic is developing its artificial intelligence systems around twice as quickly as it was a year ago amid the CEO’s calls to slow down the technology’s pace of improvement, according to an analysis released Thursday by an independent monitoring group.
Anthropic invited researchers from the Berkeley, California-based nonprofit, Model Evaluation and Threat Research, to review its internal data and interview staff as it prepares for its Wall Street debut with debate swirling over the potential for AI models to elude human control with catastrophic consequences.
The findings are significant because details about the technology — and its risks — have been provided primarily by AI companies, which have been encouraged by President Donald Trump’s administration to police themselves on matters of safety.
It was during the research group’s assessment period that Anthropic CEO Dario Amodei called for the AI industry to slow its work so safety measures can catch up.
In a Sept. 12 essay responding to mounting concerns about runaway AI, he urged the AI world to “slow the pace at which we improve the capabilities of AI models.
"""
```

Import Libraries

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

Apply NER

```python
doc = nlp(ner_text)
```

Display Entity and Tags

```python
print(f"{'Entity':<40} {'Entity Type':<15} {'Description'}")
print("-" * 85)

for ent in doc.ents:
    print(f"{ent.text:<40} {ent.label_:<15} {spacy.explain(ent.label_)}")
```
