Part 1: Tokenization and Normalization

Import the NLTK module

```python
import nltk
import sys

print(sys.executable)
print(nltk.data.path)

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download('tagsets')

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.util import ngrams 
from nltk import FreqDist
```

Select a text file or excerpt to work with (e.g. a paragraph from a book, a news article, etc.)

"""

Yet Harry Potter was still there, asleep at the moment, but not for long. His Aunt Petunia was awake and it was her shrill voice that made the first noise of the day.

```python
example_string = """Nearly ten years had passed since the Dursleys had woken up to find their nephew on the front step, but Privet Drive had hardly changed at all. The sun rose on the same tidy front gardens and lit up the brass number four on the Dursleys' front door; it crept into their living room, which was almost exactly the same as it had been on the night when Mr. Dursley had seen that fateful news report about the owls. Only the photographs on the mantelpiece really showed how much time had passed. Ten years ago, there had been lots of pictures of what looked like a large pink beach ball wearing different-colored bonnets - but Dudley Dursley was no longer a baby, and now the photographs showed a large blond boy riding his first bicycle, on a carousel at the fair, playing a computer game with his father, being hugged and kissed by his mother. 
The room held no sign at all that another boy lived in the house, too.

Yet Harry Potter was still there, asleep at the moment, but not for long. His Aunt Petunia was awake and it was her shrill voice that made the first noise of the day."""
```

Tokenize the text using nltk's word_tokenize() and sentence_tokenize() functions

```python
sent_tokenize(example_string, language="english")
```

```python
words = word_tokenize(example_string)

words
```

Normalize the tokens by converting them to lowercase and stemming using nltk's PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in words]

stemmed_words
```

Part 2: Part-of-Speech Tagging

Use nltk's pos_tag() function to tag each tokenized word with its part-of-speech

```python
nltk.pos_tag(words)
```

```python
nltk.help.upenn_tagset()
```

Examine the tagged words and take note of any patterns, surprises or errors you see

Overall, the tagging is mostly accurate for the paragraph. However, there are a few patterns of errors that I noticed like incorrect plurality and adjective vs nouns. Most of the nouns are categorized currently, however, I did find some errors where nouns like ‘Dursleys’, ‘owls’, and ‘photographs’ are marked as singular or mass nouns rather than plural nouns. There are some words like ‘asleep’, ‘blond’, and ‘tidy’ that are marked as nouns but are adjectives in the context of the paragraph. 

Part 3: N-Grams

Use ngrams() to create a series of bi-grams and tri-grams from the tokenized text

```python
bigrams = list(ngrams(words, 2))
trigrams = list(ngrams(words, 3))

print("Bigrams: ", bigrams)
print("Trigrams: ", trigrams)
```

Calculate the frequency distribution of the n-grams using FreqDist()

```python
bigram_frequency_distribution = FreqDist(bigrams)

print(bigram_frequency_distribution)
```

```python
print(bigram_frequency_distribution.most_common(20))
```

```python
trigram_frequency_distribution = FreqDist(trigrams)

print(trigram_frequency_distribution)
```

```python
print(trigram_frequency_distribution.most_common(20))
```
